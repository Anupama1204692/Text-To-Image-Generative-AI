```python
from diffusers import StableDiffusionPipeline
import torch
import gradio as gr

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
)
pipe = pipe.to("cuda")
print("Model Ready!")
```

```python
def generate_image(prompt):
    image = pipe(prompt, num_inference_steps=25).images[0]
    return image

gr.Interface(
    fn=generate_image,
    inputs=gr.Textbox(label="Prompt likho", placeholder="Varanasi ghat at sunrise"),
    outputs="image",
    title="Text to Image"
).launch()
```
